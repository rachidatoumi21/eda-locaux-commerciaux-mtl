```python
from pathlib import Path

src = Path("data/occupation-commerciale-2025.csv")
dst = Path("data/occupation-commerciale-2025-fixed.csv")

with src.open("r", encoding="utf-8", errors="replace", newline="") as fin, \
     dst.open("w", encoding="utf-8", newline="") as fout:
    
    for i, line in enumerate(fin):
        line = line.rstrip("\n\r")

        # garder l'en-tête tel quel
        if i == 0:
            fout.write(line + "\n")
            continue

        # enlever les guillemets externes s'ils enveloppent toute la ligne
        if line.startswith('"') and line.endswith('"'):
            line = line[1:-1]

        # convertir les doubles guillemets "" en " (format CSV standard)
        line = line.replace('""', '"')

        fout.write(line + "\n")

print("✅ Fichier corrigé créé :", dst)
```

```python
import pandas as pd

df = pd.read_csv("data/occupation-commerciale-2025-fixed.csv", sep=",", low_memory=False, encoding="utf-8")
df.head()
```

```python
df.iloc[0].notna().sum(), df.shape
```

```python
# 1) enlever doublons
df = df.drop_duplicates()

# 2) nettoyer les textes (espaces)
for col in ["ARRONDISSEMENT", "QUARTIER", "SECTEUR_PME", "ORIGINE", "T_COMMERCE"]:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()

# 3) transformer la date
if "DATE_CREATION" in df.columns:
    df["DATE_CREATION"] = pd.to_datetime(df["DATE_CREATION"], errors="coerce")

df.info()
```

```python
[c for c in df.columns if "VAC" in c.upper() or "OCCUP" in c.upper() or "STAT" in c.upper()]
```

```python
col = "VACANT_A_LOUER"  # change si le nom est différent
df[col].value_counts(dropna=False).head(20)
```

```python
col = "VACANT_A_LOUER"
counts = df[col].value_counts(dropna=False)
counts.plot(kind="bar")
plt.title("Répartition du statut : vacant à louer")
plt.ylabel("Nombre de locaux")
plt.show()
```

```python
vac = df[df[col].astype(str).str.lower().isin(["oui", "true", "1", "vacant"])]

top_vac = vac["ARRONDISSEMENT"].value_counts().head(10)
top_vac.plot(kind="bar")
plt.title("Top 10 arrondissements — nombre de locaux vacants")
plt.ylabel("Nombre de locaux vacants")
plt.show()
```

```python
total = df.groupby("ARRONDISSEMENT").size()
vac_n = vac.groupby("ARRONDISSEMENT").size()

rate = (vac_n / total).fillna(0).sort_values(ascending=False).head(10)
rate.plot(kind="bar")
plt.title("Top 10 arrondissements — taux de vacance")
plt.ylabel("Taux de vacance")
plt.show()
```

```python
total = len(df)
vacants = (df["VACANT_A_LOUER"] == "Oui").sum()
taux_global = vacants / total

print("Nombre total de locaux:", total)
print("Nombre vacants à louer:", vacants)
print("Taux global de vacance:", round(taux_global*100, 2), "%")
```

```python
# Top secteurs par taux de vacance (min 200 locaux pour éviter les petits groupes)
g = df.groupby("SECTEUR_PME").agg(
    total=("VACANT_A_LOUER", "size"),
    vacants=("VACANT_A_LOUER", lambda s: (s=="Oui").sum())
)
g["taux"] = g["vacants"] / g["total"]

g = g[g["total"] >= 200].sort_values("taux", ascending=False).head(10)

g["taux"].plot(kind="bar")
plt.title("Top 10 secteurs — taux de vacance (min 200 locaux)")
plt.ylabel("Taux de vacance")
plt.show()
```

```python
g = df.groupby("T_COMMERCE").agg(
    total=("VACANT_A_LOUER", "size"),
    vacants=("VACANT_A_LOUER", lambda s: (s=="Oui").sum())
)
g["taux"] = g["vacants"] / g["total"]
g = g.sort_values("taux", ascending=False).head(10)

g["taux"].plot(kind="bar")
plt.title("Top 10 types de commerce — taux de vacance")
plt.ylabel("Taux de vacance")
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1) on garde seulement les vacants
vac = df[df["VACANT_A_LOUER"] == "Oui"].copy()

# 2) compter par secteur
s = vac["SECTEUR_PME"].fillna("Inconnu").value_counts()

# 3) Top 7 + "Autres"
top_n = 7
s_top = s.head(top_n)
autres = s.iloc[top_n:].sum()
if autres > 0:
    s_top.loc["Autres"] = autres

# 4) camembert
plt.figure()
plt.pie(s_top, labels=s_top.index, autopct="%1.1f%%", startangle=90)
plt.title("Répartition des locaux vacants par secteur (SECTEUR_PME)")
plt.axis("equal")
plt.show()
```
